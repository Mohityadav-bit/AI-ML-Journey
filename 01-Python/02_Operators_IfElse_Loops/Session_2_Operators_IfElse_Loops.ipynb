## 1. Operators
### Arithmetic

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `//` | Floor division |
| `%` | Modulus |
| `**` | Power |

```python
# Comparison Operators

a = 10
b = 20

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

```python
# Logical Operators

'''
and
or
not
'''


# Example:
age = 21
has_id = True
print(age >= 18 and has_id)
```

```python
# Assignment Operators
x = 10
x += 5
print(x)
x -= 2
print(x)
x *= 2
print(x)
```

```python
# Membership Operators

'''
in
not 
in
'''


# Example:
name = "Mohit"
print("M" in name)
print("z" in name)
```

```python
# Identity Operators


'''
is
is not
'''

# These check whether two references point to the same object.
a = [1, 2]
b = a
print(a is b)


# Important:
# ==      # checks value equality.

# is      # checks object identity.
```

```python
# If-Else
# Used for decision making.
# Syntax
'''

if condition:
    statement
elif condition:
    statement
else:
    statement


'''
# Example:
age = 21
if age >= 18:    
    print("Adult")
else:    
    print("Minor")
```

```python
# Nested If
age = 21
citizen = True
if age >= 18:    
    if citizen:        
        print("Eligible")
```

```python
# Modules
# A module is a Python file containing reusable code.
# Example:
import math
print(math.sqrt(25))
print(math.factorial(5))


# You can also import specific functions:
from math import sqrt
print(sqrt(25))
```

```python
# While Loop
# Used when repetition depends on a condition.
# Syntax
'''

while condition:    
    statement


'''

# Example:
i = 1
while i <= 5:    
    print(i)    
    i += 1
```

```python
# For Loop
# Used to iterate over an iterable.
for i in range(1, 6):    
    print(i)
```
