#### Python is widely used in:

- Data Science
- Machine Learning
- Artificial Intelligence
- Automation
- Cybersecurity
- Scientific Computing
- Data Analysis

### Important characteristics

| Feature | Meaning |
|---|---|
| High-level | Easier for humans to understand |
| Interpreted | Executed through an interpreter |
| Dynamically typed | Variable type is determined at runtime |
| Object-oriented | Supports OOP |
| Open source | Freely available |
| Portable | Runs on different operating systems |

```python
## print()
# The print() function is used to display information on the screen.
# Syntax
# print(value)


# Example
print("Hello World")
print(10)
print(10 + 20)


# Output
# Hello World
# 10
# 30

# Printing multiple values
name = "Mohit"
age = 21
print(name, age)


# Output:
# Mohit 21

# sep
# Used to define the separator between multiple values.
print("Python", "Java", "C++", sep=" | ")


# Output:
# Python | Java | C++

# end
# By default, print() moves to a new line.
print("Hello")
print("World")


# Output:
# Hello
# World

# Using end:
print("Hello", end=" ")
print("World")
```

```python
# Python Data Types
# A data type defines what kind of value a variable contains.
# Python has several built-in data types.
# Main categories
'''
Python Data Types
│
├── Numeric
│   ├── int
│   ├── float
│   └── complex
│
├── Boolean
│   └── bool
│
├── Text
│   └── str
│
├── Sequence
│   ├── list
│   ├── tuple
│   └── range
│
├── Set
│   └── set
│
├── Mapping
│   └── dict
│
└── None
    └── NoneType

    '''
# Integer
age = 21
marks = 95


print(type(age))


# <class 'int'>

# Float
price = 99.99
percentage = 87.5
print(type(price))


# Complex
z = 3 + 4j
print(z)
print(type(z))


# Boolean
# Boolean has only two values:
# True & False


# Example:
is_student = True
print(type(is_student))


# String
name = "Mohit"
print(type(name))


# List
# A list stores multiple values and is mutable.
numbers = [10, 20, 30, 40]
print(numbers)
print(type(numbers))


# Tuple
# A tuple is an immutable collection.
numbers = (10, 20, 30)
print(numbers)


# Set
# A set stores unique elements.
numbers = {10, 20, 20, 30}
print(numbers)



# Dictionary
# Stores data in key-value pairs.
student = {    "name": "Mohit",    "age": 21,    "branch": "CSE-AIML"}
print(student)


# None
# None represents the absence of a value.
result = None
print(result)
print(type(result))
```
